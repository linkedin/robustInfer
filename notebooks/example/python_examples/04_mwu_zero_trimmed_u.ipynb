# Mann-Whitney U and Zero-Trimmed U Statistics

This notebook demonstrates non-parametric two-sample testing methods available in the RobustInfer Python library.




```python
import numpy as np
from scipy import stats

# Import RobustInfer
from robustinfer import zero_trimmed_u
```

## Generate Test Data

Let's create datasets with different characteristics to test the methods.


```python
# Generate zero-inflated positive data
np.random.seed(42)
n = 50

# Group 1: 60% zeros, 40% positive values from exponential distribution
zero_prob1 = 0.6
group1_indicators = np.random.binomial(1, 1 - zero_prob1, n)
group1_positive = np.random.exponential(scale=1.0, size=n)
group1 = group1_indicators * group1_positive

# Group 2: 40% zeros, 60% positive values from exponential distribution (different scale)
zero_prob2 = 0.4
group2_indicators = np.random.binomial(1, 1 - zero_prob2, n)
group2_positive = np.random.exponential(scale=1.5, size=n)
group2 = group2_indicators * group2_positive

print(f"Group 1 - Zeros: {np.sum(group1 == 0)}/{n} ({np.sum(group1 == 0)/n*100:.1f}%), Mean of positives: {group1[group1 > 0].mean():.2f}")
print(f"Group 2 - Zeros: {np.sum(group2 == 0)}/{n} ({np.sum(group2 == 0)/n*100:.1f}%), Mean of positives: {group2[group2 > 0].mean():.2f}")
```

## Apply Different Tests


```python
# Mann-Whitney U test
mw_stat, mw_pval = stats.mannwhitneyu(group1, group2, alternative='two-sided')
print(f"Mann-Whitney U:   statistic = {mw_stat:7.4f}, p-value = {mw_pval:.4f}")

# Zero-trimmed U test
ztu_stat, ztu_var, ztu_pval = zero_trimmed_u(group1, group2)
print(f"Zero-trimmed U:   statistic = {ztu_stat:7.4f}, p-value = {ztu_pval:.4f}")
```
